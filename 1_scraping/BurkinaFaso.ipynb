```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import time
from datetime import datetime
import pandas as pd
```

```python
driver = webdriver.Chrome()
driver.get("LINK") #add link
```

```python
# search for calls for tenders

search_input = driver.find_element(By.ID, "news-subject")
search_input.send_keys("appel d'offre")

submit_button = driver.find_element(By.CSS_SELECTOR, "input[type='submit']")
submit_button.click()

time.sleep(3)
```

```python
# scraping

today = datetime.today()
tenders = []

results = driver.find_elements(By.CSS_SELECTOR, "h5.title-5")
for res in results:
    title =res.text

    a_tag = res.find_element(By.TAG_NAME, "a")
    link = a_tag.get_attribute("href")

    container = res.find_element(By.XPATH, "./ancestor::div[@class='content']")
    date_element = container.find_element(By.CSS_SELECTOR, "time")
    date_text = date_element.get_attribute("datetime")
    publish_date_obj = datetime.strptime(date_text, "%Y-%m-%d")
    
    if (today - publish_date_obj).days <= 30:
        publish_date = publish_date_obj.strftime("%d/%m/%Y")
        tenders.append({
            "title": title,
            "publication_date": publish_date,
            "expiration_date": None,
            "link": link,
            "country": "Burkina Faso"
        })
print(tenders)
```

```python
driver.quit()
```

```python
# save file

df= pd.DataFrame(tenders)
df.to_excel("../2_scraped_files/BurkinaFaso.xlsx", index=False) # modify
```
